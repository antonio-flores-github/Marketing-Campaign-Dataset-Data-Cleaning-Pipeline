```python
# Calcular la media de 'Recency' y 'MntFruits'
mean_recency = df['Recency'].mean()
mean_mntfruits = df['MntFruits'].mean()

# Reemplazar valores nulos en 'Recency' con la media
initial_nulls_recency = df['Recency'].isnull().sum()
df['Recency'] = df['Recency'].fillna(mean_recency)
final_nulls_recency = df['Recency'].isnull().sum()

if initial_nulls_recency > 0:
    print(f"\nSe reemplazaron {initial_nulls_recency} valores nulos en 'Recency' con la media ({mean_recency:.2f}).")
else:
    print("\nNo se encontraron valores nulos en 'Recency' para reemplazar.")

# Reemplazar valores nulos en 'MntFruits' con la media
initial_nulls_mntfruits = df['MntFruits'].isnull().sum()
df['MntFruits'] = df['MntFruits'].fillna(mean_mntfruits)
final_nulls_mntfruits = df['MntFruits'].isnull().sum()

if initial_nulls_mntfruits > 0:
    print(f"Se reemplazaron {initial_nulls_mntfruits} valores nulos en 'MntFruits' con la media ({mean_mntfruits:.2f}).")
else:
    print("No se encontraron valores nulos en 'MntFruits' para reemplazar.")
```

```python
import numpy as np

outliers_by_column = {}

for column in df.columns:
    if df[column].dtype in ['int64', 'float64']:
        # Numerical columns: Use IQR method
        Q1 = df[column].quantile(0.25)
        Q3 = df[column].quantile(0.75)
        IQR = Q3 - Q1
        lower_bound = Q1 - 1.5 * IQR
        upper_bound = Q3 + 1.5 * IQR

        # Identify outliers
        column_outliers = df[(df[column] < lower_bound) | (df[column] > upper_bound)][[column]].drop_duplicates()

        if not column_outliers.empty:
            outliers_by_column[column] = column_outliers[column].tolist()

    elif df[column].dtype == 'object':
        # Categorical columns: Use frequency-based method (less than 1% of total rows)
        value_counts = df[column].value_counts()
        threshold = len(df) * 0.01 # 1% threshold

        low_frequency_values = value_counts[value_counts < threshold].index.tolist()

        if low_frequency_values:
            outliers_by_column[column] = low_frequency_values

print("\n--- Valores atípicos identificados por columna ---")
if outliers_by_column:
    for col, outliers in outliers_by_column.items():
        print(f"Columna '{col}': {outliers}")
else:
    print("No se identificaron valores atípicos significativos en ninguna columna.")
```

```python
# Limpiar la columna 'Income': reemplazar valores no numéricos con la media
# Convertir a numérico, forzando errores a NaN
df['Income'] = pd.to_numeric(df['Income'], errors='coerce')
# Calcular la media de la columna 'Income' (ignorando NaN)
mean_income = df['Income'].mean()
# Reemplazar NaN con la media
initial_nulls_income = df['Income'].isnull().sum()
df['Income'] = df['Income'].fillna(mean_income)
if initial_nulls_income > 0:
    print(f"Se reemplazaron {initial_nulls_income} valores no numéricos/nulos en 'Income' con la media ({mean_income:.2f}).")
else:
    print("No se encontraron valores no numéricos/nulos en 'Income' para reemplazar.")

# Limpiar la columna 'Kidhome': reemplazar 'N' con 0
initial_N_kidhome = df[df['Kidhome'] == 'N'].shape[0]
df['Kidhome'] = df['Kidhome'].replace('N', 0)
if initial_N_kidhome > 0:
    print(f"Se reemplazaron {initial_N_kidhome} valores 'N' en 'Kidhome' con 0.")
else:
    print("No se encontraron valores 'N' en 'Kidhome' para reemplazar.")

# Limpiar la columna 'Teenhome': reemplazar 'N' con 0
initial_N_teenhome = df[df['Teenhome'] == 'N'].shape[0]
df['Teenhome'] = df['Teenhome'].replace('N', 0)
if initial_N_teenhome > 0:
    print(f"Se reemplazaron {initial_N_teenhome} valores 'N' en 'Teenhome' con 0.")
else:
    print("No se encontraron valores 'N' en 'Teenhome' para reemplazar.")

# Mostrar el conteo de valores únicos para verificar
print("\nValores únicos y conteo en 'Income' después de la limpieza:")
print(df['Income'].value_counts(dropna=False))
print("\nValores únicos y conteo en 'Kidhome' después de la limpieza:")
print(df['Kidhome'].value_counts(dropna=False))
print("\nValores únicos y conteo en 'Teenhome' después de la limpieza:")
print(df['Teenhome'].value_counts(dropna=False))
```
